```python

from pathlib import Path
import zipfile
from collections import Counter

DATA = Path("datasets")
RESULTS = Path("results")
RESULTS.mkdir(exist_ok=True)

extensions = Counter()

for zip_path in DATA.glob("*.zip"):
    print("\n" + "=" * 60)
    print("Dataset:", zip_path.name)

    try:
        with zipfile.ZipFile(zip_path, "r") as archive:
            files = [
                item for item in archive.infolist()
                if not item.is_dir()
            ]

            print("Total files:", len(files))
            print("Archive size:", round(zip_path.stat().st_size / (1024**3), 2), "GB")

            for item in files:
                ext = Path(item.filename).suffix.lower()
                extensions[ext or "[no extension]"] += 1

            print("\nMost common file types:")
            for ext, count in Counter(
                Path(item.filename).suffix.lower() or "[no extension]"
                for item in files
            ).most_common(15):
                print(f"{ext}: {count}")

            report = RESULTS / f"{zip_path.stem}_contents.txt"

            with report.open("w", encoding="utf-8") as out:
                out.write(f"Dataset: {zip_path.name}\n")
                out.write(f"Total files: {len(files)}\n\n")

                for item in files:
                    out.write(
                        f"{item.filename}\t"
                        f"{item.file_size} bytes\n"
                    )

            print("\nFirst 30 files:")
            for item in files[:30]:
                print(item.filename)

            print("File list saved to:", report)

    except zipfile.BadZipFile:
        print("ERROR: This is not a valid ZIP archive.")
    except Exception as error:
        print("ERROR:", error)

print("\nInspection finished.")
```
